```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score,confusion_matrix,precision_score,recall_score,f1_score
from xgboost import XGBClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
```

```python
df=pd.read_csv('Clean_dataset.csv')
print(df.head())
```

```python
print(df.isnull().sum())
```

```python
df['TotalCharges']=pd.to_numeric(df['TotalCharges'],errors='coerce')
mean_value=df['TotalCharges'].mean()
df['TotalCharges']=df['TotalCharges'].fillna(mean_value)
print(df['TotalCharges'].isnull().sum())

```

```python
le=LabelEncoder()
for col in df.columns:
  if df[col].dtype=='object':
    df[col]=le.fit_transform(df[col])
```

```python
x=df.drop('Churn',axis=1)
y=df['Churn']
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
lr=LogisticRegression(
    max_iter=5000

)
lr.fit(x_train,y_train)
prediction_lr=lr.predict(x_test)
print(prediction_lr)
```

```python
ac_s=accuracy_score(y_test,prediction_lr)
print(ac_s)
```

```python
pre_s=precision_score(y_test,prediction_lr)
print(pre_s)
```

```python
con_m=confusion_matrix(y_test,prediction_lr)
print(con_m)
```

```python
d_tree=DecisionTreeClassifier(
    criterion='entropy',
    max_depth=5,
    random_state=42
)
d_tree.fit(x_train,y_train)
prediction_dt=d_tree.predict(x_test)
print(prediction_dt)
```

```python
ac_s=accuracy_score(y_test,prediction_dt)
print(ac_s)
```

```python
con_m=confusion_matrix(y_test,prediction_dt)
print(con_m)
```

```python
pre_s=precision_score(y_test,prediction_dt)
print(pre_s)
```

```python
f1_s=f1_score(y_test,prediction_dt)
print(f1_s)
```

```python
r_forest=RandomForestClassifier(
    n_estimators=100,
    random_state=42
)
r_forest.fit(x_train,y_train)
prediction_rf=r_forest.predict(x_test)
print(prediction_rf)
```

```python
ac=accuracy_score(y_test,prediction_rf)
print(ac)
```

```python
pre_s=precision_score(y_test,prediction_rf)
print(pre_s)
```

```python
con_m=confusion_matrix(y_test,prediction_rf)
print(con_m)
```

```python
f1_s=f1_score(y_test,prediction_rf)
print(f1_s)
```

```python
xgb=XGBClassifier(
    n_estimators=100,
    learning_rate=0.1,
    random_state=42,
    max_depth=3
)
xgb.fit(x_train,y_train)
prediction_xgb=xgb.predict(x_test)
print(prediction_xgb)
```

```python
ac=accuracy_score(y_test,prediction_xgb)
print(ac)
```

```python
pre_s=precision_score(y_test,prediction_xgb)
print(pre_s)
```

```python
con_m=confusion_matrix(y_test,prediction_xgb)
print(con_m)
```

```python
f1=f1_score(y_test,prediction_xgb)
print(f1)
```

```python
ac_lr=accuracy_score(y_test,prediction_lr)
print(ac_lr)
ac_dt=accuracy_score(y_test,prediction_dt)
print(ac_dt)
ac_rf=accuracy_score(y_test,prediction_rf)
print(ac_rf)
ac_xgb=accuracy_score(y_test,prediction_xgb)
print(ac_xgb)
```
